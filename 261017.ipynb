```python
t.initializeTurtle()      # 거북이를 초기화(세팅, 소환)

t.color("green")

for i in range(4):        # 임의로 i라는 변수에 4를 저장함, 들여써서 i 번 반복할 실행문의 내용 작성
  t.forward(160)
  t.left(90)
```

```python
t.initializeTurtle()
t.color("#00ffff")                   # 시안 색상

def draw_circle(radius):             # draw_circle 이라는 이름의 함수, radius라는 변수를 매개하는 함수를 작성(정의)하고자 함 => draw_circle(radius), f() 또는 f(a)의 형식
  for _ in range(36):
    t.forward(radius * 0.174)        # 반지름*0.174 => 중심각으로부터 10도를 이루는 호의 길이만큼 직진
    t.left(10)                       #

n = 30
t.speed(10)
for x in range(n):
  draw_circle(80)
  t.left(360 / n)
```
